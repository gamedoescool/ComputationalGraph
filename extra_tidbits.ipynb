
Consider a tensor-node with $\bold{T_{data}} = \bold{T_{input,1}} * \bold{T_{input,2}}$. How should we update the policies given $\frac{\partial \bold{E}}{\partial \bold{T_{data}}}$?

First, lets do $\bold{T_{input,2}}$:

$\frac{\partial \bold{E}}{\partial \bold{T_{input,2}}} \cdot d\bold{T_{input,2}}= \frac{\partial \bold{E}}{\partial \bold{T_{data}}} \cdot (d\bold{T_{data}}) 
= \frac{\partial \bold{E}}{\partial \bold{T_{data}}} \cdot (\bold{T_{input,1}} * d\bold{T_{input,2}})$

We are stuck now. We want to extract $d\bold{T_{input,2}}$ from this mess. 

Similarly, for $\frac{\partial \bold{E}}{\partial \bold{T_{input,1}}}$ we get:
$\frac{\partial \bold{E}}{\partial \bold{T_{input,1}}} \cdot d\bold{T_{input,1}} =  \frac{\partial \bold{E}}{\partial \bold{T_{data}}} \cdot (d\bold{T_{input,1}} * \bold{T_{input,2}})$


In general, we have the form $\bold{A} \cdot (\bold{B} * \bold{C})$ and want a way to "extract" $\bold{B}$ and $\bold{C}$. First thing to notice is that this form is trilinear, meaning its linear with respect to all of its tensors. 

Rewriting our goal in function notation, we have $f(\bold{A},\bold{B},\bold{C}) = \bold{A} \cdot (\bold{B} * \bold{C})$ and we want a $g_1(\bold{A},\bold{C})$ such that $f(\bold{A},\bold{B},\bold{C}) = g_1(\bold{A},\bold{C}) \cdot \bold{B}$.

Notice one KEY detail: f is trilinear. For F to be trilinear, then g_1 must be linear with respect to both $\bold{A}$ and $\bold{C}$. Going a step futher we can say, $g_1(\bold{A},\bold{C}) = \bold{T}\bold{A}$ for some $\bold{T}$ that is linearly dependent on $\bold{C}$. This $T$ is called the **right adjoint** of $\bold{C}$, and we will denote it as $\bold{C}^T$. 

The notation $\bold{T}\bold{A}$ deserves some comment. Here, we are really looking at a bilinear operator that naturally extends matrix multiplication to tensor-to-tensor scenes. The actual implementation isint necessary to understand right now, but just know this shorthand notation is used to denote a certain "standard" bilinear operator used to represent tensor-to-tensor linear transforms. 

Look familiar? thats because the adjoint is an extended version of the transpose we know and love. We will prove this later on however.

To be clear, this adjoint is completely dependent on its bilinear operator. the adjoint is a property of the bilinear operator at hand.

In summary, $\bold{A} \cdot (\bold{B} * \bold{C}) = (\bold{B}^T\bold{A}) \cdot {\bold{C}}$. Note that the dot product used on the RHS operates on a fundamentally different dimension then the LHS. 

Using similar linear logic, we can define the **left adjoint** $\bold{B}^{T'}$ to be a certain operation such that $\bold{A} \cdot (\bold{B} * \bold{C}) = (\bold{A}\bold{B^{T'}}) \cdot {\bold{C}}$

In general, the ordering of matracies when writing down multiplications is purely convention used to help us understand and keep track of the linear algebra madness. 

Note that adjoints must also be linear with respect to their input in order for our original expression to be trilinear. 

Also note that adjoints are invertable. THis is because of a symmetry argument: in the expression $(\bold{A}\bold{B^{T'}}) \cdot {\bold{C}}$ we can use the associated adjoints for the standard bilnear map between $\bold{A}$ and $\bold{B^{T'}}$ to reverse $\bold{B^{T'}}$ back to the right hand side.
