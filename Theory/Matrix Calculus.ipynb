**Definition of a derivative**
Consider function f(X) that takes in vector space X and outputs in another vector space. Its derivative, $f'(X)$ is the unique linear operator that acts on the two spaces such that

$$df(X) \approx f'(X)[dX]$$

For first order changes and small $dX$.

The rigorous version of this uses norms, and it can be proven that this operator is unique. However for ML intuition is fine.

The $df(X)$ is what is called a total differential. You can think of it as $df(X) = f(X+dX)-f(X)$ in vector spaces. 

Note that the general definition of a derivative includes jacobian and gradient, which we will see later.

**RULES**

**Product**
Let $X \in V_X$.
And let $T \in L: V_X \rightarrow V_Y$. i.e. T is a linear operator from the space of all linear operators from $V_X$ to $V_Y$.

Define $f(X,T) = T[X]$. Now this might be a little confusing. We have a function that takes in two inputs: $X$ from $V_X$ (domain vector space) and $T$ from $L$ (space of **all** linear operators from $V_X$ to $V_Y$). 

The total differential $df(X,T)$ is:

$$df(X,T) = (T+dT)[X+dX]-T[X] = T[dX] + dT[X] + T[X]-T[X] + dT[dX] = T[dX] + dT[X]$$

Note we ignore second level differentials, i.e. differentials that multiply each other in some sense.

This is true for **any** linear operator. Things like $a \times b$, $a \cdot b$, and scalar multiplication can be considered linear operators.

**Linearity**
The total differential is a linear operator, meaning:

$$d(af(X) + bg(Y)) = a \cdot df(X) + b \cdot dg(Y)$$

This arises naturally because of the definition of a differential and addition/subtraction properties.



**Composition rule for derivatives**
Perhaps the most powerful of them all. 
Given two functions, $f: V_X \rightarrow V_Y$ and $g: V_Y \rightarrow V_Z$ define $h(X) = g(f(X))$. 

$$dh(X) = g'(f(X))[f'(X)[dX]] \implies h'(x) = g'(f(X))f'(X)$$

Remember: g'(Y) is a function that takes in $Y$, an element of $V_Y$ and outputs the **linear operator** $L: V_X \rightarrow V_Z$ that approximates $dg$ at Y. Here, the "product" of $g'(f(X))$ and $f'(X)$ is a composition of two linear transforms. For matrices, this is matrix multiplication.

*proof*

$dh(X) = g(f(X+dX)) - g(f(X)) = g(f(X) + f'(X)[dx]) - g(f(X)) = g(f(X)) + g'(f(X))[f'(X)[dX]] - g(f(X)) = g'(f(X))[f'(X)[dX]]$

$h'(X)[dX] = g'(f(X))[f'(X)[dX]] \implies h'(X) = g'(f(X))f'(X)$

*proof end*

The notation might be confusing here: notice that dX is being put into $f'(X)$ to spit out $df$, which goes into the linear operator $g'(f(X))$ and finally a vector comes out.

In general, if you have linear transforms $A: V_X \rightarrow V_Y$ and $B: V_Y \rightarrow V_Z$ then $BA$ is used to denote the unique transform such that:

$BA[X] = B[A[X]]$

$BA$ is a linear, which makes sense as the composition of linear transformations is still linear.

Personally I believe this version of the derivative one with great potential to simplify many of the complex chain rule calculations seen in Machine Learning. Many of the derivative rules for famous models, like neural nets, are often treated as givens that can be computed with enough brute force and linear algebra pattern matching. It's honestly a waste as such derivations can be easily simplified if looked at through the right perspective. 


Now, lets connect this version of the derivative to the ones we know all too well.

First, scalars. Note that $f'(x)[dx] = f'(x)dx$ in the scalar world.

$df(x) = f(x+dx)-f(x) = f'(x)dx \rightarrow f'(x) = \frac{f(x+dx)-f(x)}{dx}$

Now, vector to scalar functions.

$df(\vec{X}) = f'(\vec{X})[d\vec{X}]$.

Note that a linear operator that goes from $\R^n$ to $\R$ is just a dot product. Thus $f'(\vec{X})[d\vec{X}] = M\cdot d\vec{x} = \nabla f \cdot d\vec{x}$

Now, the reason why each component of the gradient is just a partial is because this linear approximation must work for **all** $d\vec{x}$. In practice, this means any "direction" we can approach zero must be well behaved. So, we choose one where 

$d\vec{x} = dx_i \cdot \hat{i}$

Thus, $\nabla f \cdot (dx_i \cdot \hat{i}) =(\nabla f_i)(dx_i) =  f(x + (dx_i)\hat{i}) - f(x)$

Dividing by $dx_i$ yeilds:

$(\nabla f)_i = \frac{f(x + (dx_i)\hat{i}) - f(x)}{dx_i} = \frac{\partial f}{\partial x_i}$

Thus each component is just a partial derivative!

Similar things can be done for vec to vec functions and scalar to vec functions as well, where you can prove the global linear operator version of derivatives can be written as a matrix of partials.

Now for something interesting: what I call self similarity.

Imagine you have a function $f(X_1,X_2...,X_n)$ that takes in multiple vectors $X_i$, with vector spaces $V_i$ associated with them. **note that each input could be a vector from a different vector space from the others**

We want to find $f'(X_1,X_2...,X_n)$. What we can do is define a "container" space $X = [X_1, X_2, X_3..., X_n]$ that contains all the vector spaces, and has addition and scalar multiplication defined element wise. Note that this forms a proper vector space.

Then, the derivative $f'(X_1,X_2...,X_n)$ takes in $[dX_1, dX_2, dX_3..., dX_n]$ and outputs $df$.

Lets $T = f'(X_1,X_2...,X_n)$ for clarity.

Then, we can write $dX = [dX_1, dX_2, dX_3..., dX_n]$ into a "basis" of sorts:

$$dX = [dX_1, dX_2, dX_3..., dX_n] = [dX_1,0,0..,0] + [0,dX_2,0..,0] + ... = \sum_{i=1}^n e_i[dX_i]$$

Where $e_i$ is the linear operator that takes in $dX_i \in V_X$ and outputs $[0,0...,dX_i...,0]$ (every element 0 except at ith index).

Then: $T[\sum_{i=1}^n e_i[dX_i]] = \sum_{i=1}^n T[e_i[dX_i]]$

Now define $M_i = Te_i$, using the composition of linear operators rule.

$\sum_{i=1}^n T[e_i[dX_i]] = \sum_{i=1}^n M_i[dX_i]$.

What **is** $M_i$? Well, we use the same trick used to match the gradient to partial derivatives.

Let $[dX_1, dX_2, dX_3..., dX_n] = [0,0...,dX_c...,0]$ where $dX_c != 0$ 

This gives:

$df = T[dX] = \sum_{i=1}^n M_i[dX_i] = M_c[dX_c]$

Now note that $df = f(X + dX) - f(X) = f([X_1,X_2,...,X_c + dX_c,...,X_n]) - f([X_1,X_2...,X_n])$

Note that now $df$ is the quantity that you get when considering a differential that only acts on $X_c$. Thus $df = f'(X_c)[dX_c]$, where $f'(X_c)$ denotes the derivative of the function assuming only one input changes.

Therefore: $M_c[dX_c] = f'(X_c)[dX_c] \implies M_c = f'(X_c)$

Overall: $df = \sum_{i=1}^n f'(X_i)[dX_i]$

This is exactly the rule of adding gradients that is derived for graph neural nets. In graph neural networks, each layer can output to multiple other layers. To find the total gradient (derivative of a scalar function with respect to multiple inputs), we need to add up the individual gradients that come from each outgoing edge as per this formula. 

In this situation, we realize that the cost function $f(W_1,W_2...)$ has to have the same input weights, as the weights are shared across all outgoing edges. This simplifies the formula and allows us to directly add the operators up

$df = \sum_{i=1}^n f'(X_i)[dX] = (\sum_{i=1}^n f'(X_i))[dX] \implies f'(X) = (\sum_{i=1}^n f'(X_i))$

This means the total gradient for that node is just the sum of the ingoing gradients!
